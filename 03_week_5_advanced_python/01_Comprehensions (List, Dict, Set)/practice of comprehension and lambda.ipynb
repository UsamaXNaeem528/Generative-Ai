```python
#  Exercise 2: Lambda + Map (The Transformers)
# Task: You have a list of prices with dollar signs: prices = ["$5.99", "$12.50", "$3.45", "$99.99"].

# Using map() and a lambda function, remove the $ sign and convert them all to float numbers.

# (Bonus): Use map() to round those floats to the nearest whole integer (using the built-in round() function—no lambda needed for the bonus, just pass the function!).
```

```python
prices = ["$5.99", "$12.50", "$3.45", "$99.99"]
```

```python
prices = list(map(lambda x : x.replace('$', ''), prices))
```

```python
prices = list(map(lambda x : float(x), prices))
```

```python
prices
```

```python
prices = list(map(lambda x : round(x), prices))
```

```python
prices
```

```python
# Exercise 3: Lambda + Filter (The Gatekeepers)
# Task: You have a list of user ages: ages = [15, 22, 18, 34, 12, 45, 16, 67, 19].

# Using filter() and a lambda, keep only the users who are 18 or older (adults).

# Using filter() and a lambda, keep only the users who are between 20 and 40 (inclusive).
```

```python
ages = [15, 22, 18, 34, 12, 45, 16, 67, 19]
```

```python
age_elder = list(filter(lambda x : x>=18, ages))
```

```python
age_user = list(filter(lambda x : x>=20 and x<=40, ages))
```

```python
# Exercise 4:
# Task: You have two lists:
# students = ["Alice", "Bob", "Charlie", "Diana"]
# scores = [88, 72, 95, 83]

# Use zip() to combine them into a list of tuples: [('Alice', 88), ('Bob', 72), ...].

# (Harder): Use zip() inside a dictionary comprehension to create a single dictionary where the student's name is the key and the score is the value.
```

```python
students = ["Alice", "Bob", "Charlie", "Diana"]
scores = [88, 72, 95, 83]

my_zip = zip(students, scores)
print(list(my_zip))
```

```python
# new_dict = {key_expression: value_expression for item in iterable}
```

```python
#  Exercise 5: Enumerate (The Index Tracker)
# Task: You have a list: fruits = ["apple", "banana", "cherry", "date", "elderberry"].

# Loop through it using enumerate() and print: "Index 0: apple", "Index 1: banana", etc.

# Use enumerate() to find the index position of the fruit "cherry" (without using .index()!).
```

```python
# Exercise 6:
# The Scenario: You have a raw messy string from a website:
# raw_data = "apple:5, banana:3, CHERRY:  7 , date:2, elderberry: 10"
# Filter out any fruit where the number is less than 4.

# Your Mission (Final Output should be):
# {'apple': 5, 'cherry': 7, 'elderberry': 10}
```

```python
raw_data = "apple:5, banana:3, CHERRY:  7 , date:2, elderberry: 10"
```

```python
fruit_list = raw_data.split(',')
```

```python
clean_fruits = []
```

```python
for fruit in fruit_list:
  clean_fruits.append(fruit.replace(' ','').lower())
```

```python
clean_fruits
```

```python
fruit_dict = {}
```

```python
parts = []
```

```python
for fruit in clean_fruits:
  name, quan=fruit.split(':')
  quan = int(quan)
  parts.append((name,quan))
```

```python
parts
```

```python
for name, quan in parts:
  if quan >=4:
    fruit_dict[name] = quan
```

```python
fruit_dict
```

```python
new_fruits = {fruit: quan for fruit, quan in parts if quan>=4}
```

```python
new_fruits
```
