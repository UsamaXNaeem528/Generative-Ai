```python
import pandas as pd
import numpy as np
import seaborn as sns

data = pd.read_csv('Titanic-Dataset.csv')
data.info()
```

```python
# Check the missing the values in columns and sum them.
data.isnull().sum().sort_values(ascending=False)
```

```python
# drop the cabin columns because it contains the large number of missing values.
data =  data.drop('Cabin', axis=1)
data
```

```python
# fill the missing value in age column with mean of its column.
data['Age'] = data['Age'].fillna(data['Age'].mean())
```

```python
# now replace the missing value in embarked column, with first mode of its column
data['Embarked'] =  data['Embarked'].fillna(data['Embarked'].mode())[0]
```

```python
data.isnull().sum()
```

```python
# Now the data is free from the missing value
```

<h3>Handle missing values using the KNN algorithm</h3>

```python
import pandas as pd
data=pd.read_csv('Titanic-Dataset.csv')
data.isnull().sum().sort_values(ascending=False)
```

```python
# handle the missing the values of age column using the knn-scikit-learn
# import the knn imputer
from sklearn.impute import KNNImputer

# make the knn imputer object
imputer = KNNImputer(n_neighbors=4)

# impute the missing values with knn imputer on age column.
data['Age'] = imputer.fit_transform(data[['Age']])

# check the number of missing value in age column, after the imputing
data.isnull().sum().sort_values(ascending=False)
```

<h3>Handle Missing Values using the Iterative Imputer Scikit Learn</h3>

```python
import pandas as pd
import numpy as np
import seaborn as sns
## 1. Explicitly require the experimental feature
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df = sns.load_dataset('titanic')
df
```

```python
df.isnull().sum().sort_values(ascending=False)
```

<h5>Before imputation convert the categorical columns to the Numerical, iterative imputer expects</br>numerical data</h5>

```python
from sklearn.preprocessing import LabelEncoder

# columns to encode
columns_to_encode = ['sex','embarked','class','who','deck','embark_town','alive']

# dictionary to store Labels Encode value
label_encoders = {}

for col in columns_to_encode:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[[col]])
    label_encoders[col] = le

df.head()
```

```python
label_encoders
```

```python
# now after converting and ensuring all data in numerical for impute missing values

# Initialize and apply the imputer
imputer = IterativeImputer(max_iter=10)
X_imputed = imputer.fit_transform(df)  # this will return the numpy array
print(type(X_imputed))
```

```python
print(X_imputed)
```

```python
# For Get the column names back, just create a DataFrame from the numpy array
df_imputed = pd.DataFrame(
    X_imputed,
    columns=df.columns,
    index=df.index
)

df_imputed.head(5)
```

```python
#check again for the missing values
df_imputed.isnull().sum().sort_values(ascending=False)
```

```python
# Now the data is free from missing value using the scikit learn iterative imputer
```

```python
# we cant inverserse the encoded numerical value back to categorical

for col in columns_to_encode:
    le = label_encoders[col]
    df[col] = le.inverse_transform(df[[col]])

df.head()
```
