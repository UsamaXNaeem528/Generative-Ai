<h1>Handling Missing Values Using Pandas</h1>

<h3>Detecting a missing value in Data Frame</h3>

```python
'''
np.isnull()
return 
TRUE for missing values, 
FALSE for non-missing values means value is actually present.

Missing value can be in the column or data:
1- NaN 
2- None
'''
```

```python
import pandas as pd
data = {
    "Name": [
        "Alice Smith",
        None,
        "Charlie Brown",
        "Diana Prince",
        "Evan Wright",
        "Fiona Gallagher",
        "George Clark",
        "Hannah Abbott",
        "Ian Malcolm",
        "Julia Roberts"
    ],
    "Age": [28, None, 25, 41, 29, 32, 45, 23, 38, 50],
    "Salary": [65000, None, 54000, 95000, 68000, 72000, 110000, 48000, 87000, 125000],
    "PerformanceScore": [4.2, None, 4.5, 4.9, 3.5, 4.0, 4.7, 3.9, 4.1, 4.8]
}

df = pd.DataFrame(data)
df
```

```python
print(df.isnull())
print(df.isnull().sum())
```

<h3>Drop the row or column that contains the missing value</h3>

```python
'''
use df.dropna(axis=0 Or 1, inplace=True Or False)
'''
import pandas as pd
data = {
    "Name": [
        "Alice Smith",
        None,
        "Charlie Brown",
        "Diana Prince",
        "Evan Wright",
        "Fiona Gallagher",
        "George Clark",
        "Hannah Abbott",
        "Ian Malcolm",
        "Julia Roberts"
    ],
    "Age": [28, None, 25, 41, 29, 32, 45, 23, 38, 50],
    "Salary": [65000, None, 54000, 95000, 68000, 72000, 110000, 48000, 87000, 125000],
    "PerformanceScore": [4.2, None, 4.5, 4.9, 3.5, 4.0, 4.7, 3.9, 4.1, 4.8]
}

df = pd.DataFrame(data)
df
```

```python
df.dropna(axis=0, inplace=True)
df
```

```python
'''If the missing values are frequent in data then it is not good practice to drop the rows, 
this can cause the data loss
so, replace these missing values with some default values, or calculated values. A good approach.
'''
```

Replace the missing values instead of droping them

```python
'''df.fillna(value, inplace) used to replace the missing value with calculated or default value'''
'''
use df.dropna(axis=0 Or 1, inplace=True Or False)
'''
import pandas as pd
data = {
    "Name": [
        "Alice Smith",
        None,
        "Charlie Brown",
        "Diana Prince",
        "Evan Wright",
        "Fiona Gallagher",
        "George Clark",
        "Hannah Abbott",
        "Ian Malcolm",
        "Julia Roberts"
    ],
    "Age": [28, None, 25, 41, 29, 32, 45, 23, 38, 50],
    "Salary": [65000, None, 54000, 95000, 68000, 72000, 110000, 48000, 87000, 125000],
    "PerformanceScore": [4.2, None, 4.5, 4.9, 3.5, 4.0, 4.7, 3.9, 4.1, 4.8]
}

df = pd.DataFrame(data)
df
```

```python
df['Name'] = df['Name'].fillna('Unknown')
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['PerformanceScore'] = df['PerformanceScore'].fillna(0)
df
```

```python
df['Age'] = df['Age'].astype(int)
df['Salary'] = df['Salary'].round(2)
df
```

<h3>Interpolation</h3>

```python
'''Interpolation is a way of estimating a missing value by looking at the values around it.
1- Linear Interpolation (estime missing value by looking before and after values).
2- Polynomial Interpolation (use when data grows very fast or increasingly with  and non linear relation, and curved data.)
3- Time Interpolation (estime a missing value base on time closest intervals between the observation)
'''
```

```python
'''Linear Interpolation'''
import pandas as pd
data = {
    'Time' : [1, 2, 3, 4, 5],
    'Speed' : [10, 20, None, 40, 50]
}

df = pd.DataFrame(data)
print('Data with missing Values')
print(df)

print('\nApply the Linear Interpolation')
df = df['Speed'].interpolate(method='linear')
print(df)
```

```python
'''Polynomial Interpolation'''
import pandas as pd
data = {
    'Time' : [1, 2, 3, 4, 5],
    'Speed': [0, 10, None, 90, 160]
}

df = pd.DataFrame(data)
print('Data with missing Values')
print(df)

print('\nApply the Binomial Interpolation')
df = df['Speed'].interpolate(method='polynomial', order=2)
print(df)
```
