<h1>Grouping Data Pandas</h1>
Make a group of sama data, Like in data you have the Departments column, you can group the pandas frame acc. to Departments.

```python
import pandas as pd

data = {
    'Name' : ['John', 'Dal', 'Bob', 'Harry', 'Rahul', 'Hassan'],
    'Department' : ['HR', 'IT', 'IT', 'Finance', 'HR', 'Finance'],
    'Age' : [24, 45, 37, 27, 23, 24]
}

df = pd.DataFrame(data)
print(df)

# group this data frame by Department.
grouped = df.groupby("Department")['Name'].count()

print(f'\n"Grouped by department and calculate how many department in each column"')
print(grouped)
```
