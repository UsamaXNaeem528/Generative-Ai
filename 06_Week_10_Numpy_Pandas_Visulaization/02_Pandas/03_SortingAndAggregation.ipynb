```python
'''Sort DataFrame By one Column'''
import pandas as pd

data = {
    'Name' : ['Alice', 'Smith', 'Jane', 'Bob'],
    'Age' : [23, 45, 30, 28]
}
df = pd.DataFrame(data)
print(df)

print('Sorted By Age Ascending')
df.sort_values(by='Age', ascending=True, inplace=True)
print(df)
```

```python
'''Sort DataFrame By one or more Column'''
import pandas as pd
data = {
    'Name' : ['Alice', 'Smith', 'Jane', 'Bob'],
    'Age' : [28, 45, 30, 28],
    'Salary' : [60000, 50000, 35000, 45000]
}

df = pd.DataFrame(data)
df
print(df)

print('Sorted By Age and Column')
df.sort_values(by=['Age', 'Salary'], ascending=[True,False], inplace=True)
print(df)

#if the two person has same Age, then use the salary for order which person comes first. 
```

```python
'''Sort DataFrame By two or more Column'''
'''Task is that Sort by Department, Salary, Age'''
import pandas as pd
data = {
    'Name' : ['Alice', 'Smith', 'Jane', 'Bob'],
    'Age' : [28, 45, 30, 33],
    'Salary' : [60000, 50000, 35000, 50000],
    'Department' : ['HR', 'IT', 'HR', 'IT']
}

df = pd.DataFrame(data)
print(df)

print('Sorted By Age and Column')
df.sort_values(by=['Department', 'Salary', 'Age'], ascending=[True, False, False], inplace=True)
print(df)
```
