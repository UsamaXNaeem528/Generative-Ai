<h1>Grouping Data Pandas</h1>
Make a group of sama data, Like in data you have the Departments column, you can group the pandas frame acc. to Departments.

```python
import pandas as pd

data = {
    'Name' : ['John', 'Dal', 'Bob', 'Harry', 'Rahul', 'Hassan'],
    'Department' : ['HR', 'IT', 'IT', 'Finance', 'HR', 'Finance'],
    'Age' : [24, 45, 37, 27, 23, 24]
}

df = pd.DataFrame(data)
print(df)

# group this data frame by Department.
grouped = df.groupby("Department")['Name'].count()

print(f'\n"Grouped by department and calculate how many department in each column"')
print(grouped)
```

```python
import pandas as pd

data = {
    "Employee": [
        "Ali", "Sara", "John", "Ayesha", "David",
        "Hamza", "Emma", "Usman", "Sophia", "Bilal",
        "Ahmed", "Maria"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "Finance", "IT", "Marketing", "Marketing", "Finance",
        "HR", "IT"
    ],

    "City": [
        "Lahore", "Karachi", "Lahore", "Islamabad", "Karachi",
        "Lahore", "Islamabad", "Karachi", "Lahore", "Islamabad",
        "Lahore", "Karachi"
    ],

    "Age": [
        25, 30, 28, 35, 27,
        32, 24, 29, 31, 26,
        33, 23
    ],

    "Salary": [
        60000, 55000, 75000, 80000, 65000,
        72000, 58000, 62000, 70000, 68000,
        59000, 67000
    ],

    "Experience": [
        2, 5, 4, 8, 3,
        6, 1, 4, 7, 3,
        6, 2
    ]
}

df = pd.DataFrame(data)
print(df)

print(f'\n Average Salary of each Department.')
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
```
