```python
# implemtenting the boolean condition on single and multiple columns.
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Ayesha", "David",
                 "Hamza", "Emma", "Usman", "Sophia", "Bilal"],
    
    "Age": [22, 25, 28, 21, 30, 24, 27, 23, 29, 26],
    
    "Gender": ["Male", "Female", "Male", "Female", "Male",
                   "Male", "Female", "Male", "Female", "Male"],
    
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Multan",
                 "Karachi", "Lahore", "Islamabad", "Multan", "Karachi"],
    
    "Salary": [45000, 55000, 70000, 40000, 85000,
                   50000, 65000, 48000, 75000, 60000],
    
    "Department": ["IT", "HR", "Finance", "IT", "Marketing",
                       "Finance", "HR", "IT", "Marketing", "Finance"],
    
    "Experience": [1, 3, 5, 1, 7, 2, 4, 2, 6, 3],

    "PerformanceScore" : [40,45,50,55,60,65,70,80,85,90]
}

df = pd.DataFrame(data)
```

```python
"""filter those employees data their salary is greater than 50,000"""
data = df[df['Salary'] >= 50000]
data
```

```python
'''filter those employees data their salary is greater than or equal to 50,000 
and age is greater than or equal to 30'''
data = df[(df['Salary'] >=50000) & (df['Age']>=30)]
data
```

```python
'''
Using OR condition
filter such employess their salary is greater than 50,000 or Performance score is also greater than 50
'''
data = df[(df['Salary'] >= 50000) | (df['PerformanceScore'] >= 50)]
data
```
