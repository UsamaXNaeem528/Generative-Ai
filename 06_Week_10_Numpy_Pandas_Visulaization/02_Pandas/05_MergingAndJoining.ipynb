```python
import pandas as pd
```

```python
df_employees = pd.DataFrame({
    'emp_id': [101, 102, 103, 104],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'dept': ['HR', 'IT', 'IT', 'Finance']
})

# Right DataFrame: Employee salaries and locations
df_salaries = pd.DataFrame({
    'emp_id': [102, 103, 104, 105],
    'salary': [70000, 85000, 90000, 65000],
    'location': ['New York', 'Chicago', 'Austin', 'Remote']
})
```

```python
pd.merge(df_employees, df_salaries, how='cross')
# '''how = 'left / right / inner / outer / cross (m x n )'''
```
