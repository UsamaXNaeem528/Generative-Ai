gcsfs is a Python library that allows you to read and write files stored in Google Cloud Storage (GCS)

```python
'''How to save the data in file (csv,xlsx, json etc.) after cleaning , manipulation of data. Let see'''
import pandas as pd

data = {
    'Name': ['John', 'Bob', 'Dal'],
    'Age' : [10, 20, 30],
    'City' : ['Newyork', 'Chicago', 'New Jersy']
}

df = pd.DataFrame(data)
df
df.to_csv(r"06_Week_10_Numpy_Pandas_Visulaization\02_Pandas\output.csv", index=False)  # index=False Means don't add the idnex column
df.to_excel(r"06_Week_10_Numpy_Pandas_Visulaization\02_Pandas\output.xlsx", index=False)  # index=False Means don't add the idnex column
df.to_json(r"06_Week_10_Numpy_Pandas_Visulaization\02_Pandas\output.json", index=False, indent=4)  # index=False Means don't add the idnex column
```
