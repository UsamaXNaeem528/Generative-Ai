<h1>Reshaping of Numpy Array</h1>

```python
import numpy as np

# Reshaping changes the shape (dimensions) of an array without changing its data.
# For example:
# - 1D → 2D
# - 2D → 1D
# - 2D → 3D (if the number of elements remains the same)
#
# reshape() returns a reshaped array. In most cases, it returns a view of the
# original array, so both arrays share the same underlying data.

'''
reshape(new_shape)

Rules:
1. The total number of elements must remain the same.
   Example:
   Original array: 6 elements
   reshape(2, 3) -> 2 × 3 = 6 ✓
   reshape(3, 2) -> 3 × 2 = 6 ✓
   reshape(4, 2) -> 4 × 2 = 8 ✗ (raises ValueError)

2. reshape() does not modify the original array's shape.
   It returns a new array with the requested shape.

3. If the reshaped array is a view (most common case),
   changes made through either array are reflected in the other
   because they share the same underlying memory.
'''

arr = np.array([1, 2, 3, 4, 5, 6])
print("Original array:")
print(arr)

reshaped_arr = arr.reshape(2, 3)
print("\nReshaped array:")
print(reshaped_arr)

# Modify the reshaped array
reshaped_arr[0, 0] = 100
print("\nAfter modifying reshaped_arr:")
print(reshaped_arr)

# Original array is also affected
print("\nOriginal array:")
print(arr)

# Modify the original array
arr[2] = 98
print("\nAfter modifying original array:")
print(arr)

# Reshaped array reflects the change
print("\nReshaped array:")
print(reshaped_arr)
```

<h1>Stacking In Numpy</h1>

```python
'''
In NumPy, stacking means combining multiple arrays into a single array by adding them along a specified axis.
It is commonly used to join arrays either vertically, horizontally, or along a new dimension.

vstack() --> row wise
hstack() --> columns wise
'''
print('With 1D implementation\n')
arr1 = np.array([1,2,3])
arr2 = np.array([7,8,9])
 
print('Vertical stacking means rows wise')
print(np.vstack((arr1, arr2)))

print('\nHorizontal stacking means column wise')
print(np.hstack((arr1, arr2)))
```

```python
# With 2D
# Remember this
# vstack → add more rows (go downward ⬇️) vstack() places one array below the other.
# hstack → add more columns (go rightward ➡️) hstack() places one array to the right of the other.

print('With 1D implementation\n')
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = np.vstack((A, B))
print(result)

result = np.hstack((A, B))
print(result)
```

<h1>Broadcasting Numpy</h1>
<p>The most important concept in numpy</p>

```python
'''
Broadcasting is NumPy's way of automatically expanding the smaller array,
So that it can perform element-wise operations with a larger array.

Broadcasting Rules:
NumPy compares the shapes of two arrays from the last dimension (rightmost) to the first (leftmost).

Two dimensions are compatible if:
1-They are equal, or
2-One of them is 1.

If neither condition is true, broadcasting fails with a ValueError.
'''
```

```python
'''Example 1: Same Shape'''
import numpy as np

arr1 = np.array([[10,20],
                [40,50]])
arr2 = np.array([[60,70],
                [80,90]])

'''
shape --> arr1 : (2,2)
shape --> arr2 : (2,2)

Now compare the shapes of arr1 and arr2 from rightmost 
(2,2) == (2,2)  both arrays have same dimensions 2'rows and 2'rows , 1st rule is apply, so broadcasting in possible.
'''

print(arr1 + arr2)
```

```python
'''# Example 2: One Array Dimension is 1'''
A = np.array([[1,2,3],
             [4,5,6]])

B = np.array([10,20,30])

'''
shape --> A : (2,3)
shape --> B : (3)
Numpy treats shape B as the (1,3)

Now compare the shapes of A and B from right most
(2,3)
(1,3)
3==3, 2 vs 1 --> for this rule 2 is applied (here is the one of them is equal to 1)
so 'B' is broadcast to
[[10 20 30]
[10 20 30]]
'''

print(A + B)
```

```python
'''Example 3: Column Broadcasting'''
A = np.array([[1,2,3],
             [4,5,6]])

B = np.array([[10],
             [20]])

'''
shape --> A : (2,3)
shape --> B : (2,1)

Now compare the shapes of A and B from right most
(2,3)
(2,1)
2==2, 3 vs 1 --> for this rule 2 is applied (here is the one of them is equal to one)
so 'B' is broadcast to
[[10 10 10]
[20 20 20]]
'''

print(A + B)

```

```python
'''Example 4: Broadcasting Failure'''
A = np.ones((2,3))
B = np.ones((2,2))

'''
shape --> A : (2,3)
shape --> B : (2,2)

Now compare the shapes of A and B from right most left most
(2,3)
(2,2)
3==2, from right most dimension both rules are failed *Neither dimension is equal nor 1.*
so we get the broadcasting error
'''
print(A+B)
```

Handling the infinite Values in data set

```python
''' 
first how to identify the infinite values(positve and negative) in the data set
 Syntax:
 -np.isinf(arr)
'''

import numpy as np
arr = np.array([1,2,np.inf, 4, 5, np.inf])
print(np.isinf(arr))
```

```python
'''
How to replace the infinte values with numerical 
Syntax: np.nan_to_num(arr, posinf=1000, neginf=-1000)   
- arr means array contains the infinite value
- posinf means replace positive infinte value with the positive +1000
- neginf means replace neg infinte value with the neg -1000
- these value can change acc. to need not a rule
'''

import numpy as np
arr = np.array([1,2,np.inf, 4, 5, -np.inf])
# arr = np.array([1,2,3, 4, 5, 6])

found = False

if np.any(np.isinf(arr)):
    found = True
    print('Array contains the Infinte value or values')
else:
    print('Clear, No infinite value')

if found is True:
    clean_arr = np.nan_to_num(arr, posinf=100, neginf=-100)
    print(clean_arr)
```
