<h1> Aggregation Functions in python </h1>

```python
# Mean==>The average value of the data
# Variance==>How spread out the data is (in squared units) from mean
# Standard Deviation==>How spread out the data is (in the original units) from means

# Simple analogy
# Imagine students' test scores:
# Mean = the class average.
# Variance = how different the students' scores are from the average.
# Standard deviation = the average amount by which a student's score differs from the class average, expressed in the same units (points) as the scores.
```

```python
marks_arr = np.array([[48, 49, 50, 50, 51, 52]])
```

```python
print(marks_arr.sum())
print(marks_arr.mean())
print(marks_arr.min())
print(marks_arr.max())
print(marks_arr.var())
print(marks_arr.std())
```

<h1>Slicing In Numpy and it different techniques</h1>

```python
# Numpy array slicing syntax
# array[start:stop:step]
# array[start:end] , start to end -1 means the last element is excluded 
# neagtive step, -1 reverse --> Negative indexing in NumPy works exactly like Python lists: it counts from the end of the array.
import numpy as np
arr = np.array([10, 20, 30, 40, 50, 60, 70])
print(arr[0:5])
print(arr[0::2])
print(arr[2:5:2])

# [10, 20, 30, 40, 50, 60, 70] == [-6 -5 -4 -3 -2 -1]
#negative indexing 
print(arr[-1::-2])

# interview question reverse the numpy without using the loop
arr_interview = np.array([1,3,5,7,9,11,13,15])
arr_interview[-1::-1]
```

<h1>Reshaping of Numpy Array</h1>

```python
import numpy as np

# Reshaping changes the shape (dimensions) of an array without changing its data.
# For example:
# - 1D → 2D
# - 2D → 1D
# - 2D → 3D (if the number of elements remains the same)
#
# reshape() returns a reshaped array. In most cases, it returns a view of the
# original array, so both arrays share the same underlying data.

'''
reshape(new_shape)

Rules:
1. The total number of elements must remain the same.
   Example:
   Original array: 6 elements
   reshape(2, 3) -> 2 × 3 = 6 ✓
   reshape(3, 2) -> 3 × 2 = 6 ✓
   reshape(4, 2) -> 4 × 2 = 8 ✗ (raises ValueError)

2. reshape() does not modify the original array's shape.
   It returns a new array with the requested shape.

3. If the reshaped array is a view (most common case),
   changes made through either array are reflected in the other
   because they share the same underlying memory.
'''

arr = np.array([1, 2, 3, 4, 5, 6])
print("Original array:")
print(arr)

reshaped_arr = arr.reshape(2, 3)
print("\nReshaped array:")
print(reshaped_arr)

# Modify the reshaped array
reshaped_arr[0, 0] = 100
print("\nAfter modifying reshaped_arr:")
print(reshaped_arr)

# Original array is also affected
print("\nOriginal array:")
print(arr)

# Modify the original array
arr[2] = 98
print("\nAfter modifying original array:")
print(arr)

# Reshaped array reflects the change
print("\nReshaped array:")
print(reshaped_arr)
```
