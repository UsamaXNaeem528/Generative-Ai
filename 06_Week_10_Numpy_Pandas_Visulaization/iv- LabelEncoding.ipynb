```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset('tips')
df
```

<h1>One Hot Encoding</h1>

```python
# Use One hot encoding for Sex Column
# df['sex'].value_counts()
# df['sex'].isnull().value_counts()

from sklearn.preprocessing import LabelEncoder
le_encoder = LabelEncoder()
df['sex_encoded'] = le_encoder.fit_transform(df['sex'])
df
```

```python
# df['smoker'].value_counts()
# df['smoker'].isnull().value_counts()

from sklearn.preprocessing import LabelEncoder

le_encoder = LabelEncoder()
df['encoded_smoker'] = le_encoder.fit_transform(df['smoker'])
df
```

<h3>Ordinal Encoding</h3>

```python
from sklearn.preprocessing import OrdinalEncoder

day_order = ['Mon', 'Tue', 'Wed', 'Thur', 'Fri','Sat', 'Sun']
ord_encoder = OrdinalEncoder(categories=[day_order])
df['encoded_day'] = ord_encoder.fit_transform(df[['day']])
df
```

</h3> One Hot Encoding </h3>

```python
from sklearn.preprocessing import OneHotEncoder

df['embarked'].isnull().value_counts()
```

```python
# first fills the null value in embarked column using Mode
firstmode_value = df['embarked'].mode()[0]
df['embarked']=df['embarked'].fillna(firstmode_value)
```

```python
df['embarked'].isnull().value_counts()
```

```python
encoder = OneHotEncoder(sparse_output=False)

encoded_embarked = encoder.fit_transform(df[['embarked']])
encoded_embarked = pd.DataFrame(
    encoded_embarked,
    columns=encoder.get_feature_names_out(['embarked'])
)

encoded_embarked
```
