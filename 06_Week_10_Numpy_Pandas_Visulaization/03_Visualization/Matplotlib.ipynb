<h3>Sub Plotting Matplotlib</h3>
<p>Sub Plot Means Creating Multiple Plots in A single Canvas OR Figure</p>

<h2>Subplotting using plt.subplot</h3>

```python
plt.subplot(2,2,1)
plt.plot(xaxis, yaxis)

plt.subplot(2,2,4)
plt.plot([1,2,3,4,5,6],[1,2,3,4,5,6])

plt.subplot(2,2,2)
plt.plot(xaxis, yaxis)

plt.subplot(2,2,3)
plt.plot([2,3,4,5,6,7,8],[120,180,240,300,360,420,480])
plt.xlabel("Minutes")
plt.ylabel("Seconds")
```

"This sets up 4 subplots in a 2x2 grid. This is generally enough for visualization, but more subplots can be added if necessary."

<h3>Sub Plotting Using the plt.figure</h3>

plt.figure() creates a new figure (plotting window or canvas).

Think of a figure as the entire area where your plots will be displayed.

If you use plt.figure() multiple times, plots can appear on top of each other if you don't create a new figure before plotting.

Let's understand the difference.

```python
# 1. Without creating a figure both plots will appear on same axis
import matplotlib.pyplot as plt

plt.plot([1,2,3],[4,5,6])
plt.plot([1,2,3],[6,5,4])
```

```python
# With Creating a figure before each plot. Then 2 different images or figure appears.
# dpi stands Dots Per inch determines the resolution or sharpness of your figure.
plt.figure(figsize=(8,6), dpi=100)
plt.plot([1,2,3],[4,5,6])
plt.show()

plt.figure(figsize=(12,8), dpi=200)
plt.plot([1,2,3],[6,5,4])
plt.show()
```

<h3>Sub Plotting Using the plt.subplots()</h3>

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].plot([1, 2, 3], [4, 5, 6])
ax[0].set_title("First Plot")

ax[1].plot([1, 2, 3], [6, 5, 4])
ax[1].set_title("Second Plot")

plt.tight_layout()
plt.show()
```

<h3> Distribution Plots</h3>

```python
# Scatter Plots 
# Type of grahp used to show the relationship between the two numerical variable using the
# invidual dots
```

```python
hours = [1,2,3,4,5]
marks = [20,40,60,80,100]
plt.scatter(hours,marks)
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("Marks According to Hours Studied")
```

```python
# Hist Plot , plt.hist(data) which creates a histogram to show the numerical distribution of data. 
x = np.array([100,120,130,140,150,160,170,180,190,200])
plt.hist(x)
```

```python
# Box Blot OR Whisker Plot is used to understand the distribution of numerical data. It helps us to identify
#  the median, quartiles, spread of data, and outliers.

import matplotlib.pyplot as plt

marks = [40, 45, 50, 55, 60, 65, 70, 75, 80, 95, 150]

plt.boxplot(marks)

plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")

plt.show()
```

```python
# Saving the Plot Images
plt.figure(figsize=(8,6), dpi=100)
plt.scatter([1,2,3,4,5],[5,4,3,2,1])
plt.savefig("scatter-plot.png")
plt.show()
```
