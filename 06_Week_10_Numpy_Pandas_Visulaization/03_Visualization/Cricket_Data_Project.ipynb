```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Cricket.csv')
df.head()
```

```python
df.info()
```

```python
# 1. Which Team won the most matches
df['match_winner']
```

```python
df['match_winner'].value_counts()
```

```python
winner_count = df['match_winner'].value_counts()
sns.barplot(x=winner_count.values, y=winner_count.index)
```

```python
# 2.Toss Decision Trends
sns.countplot(data=df, x='toss_decision', palette='muted')
```

```python
df.head()
```

```python
# 3. How many times did the toss winner also win the match?
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100) / df.shape[0]
print(f'Percentage of team who wins the toss & also win the match : {round(percentage,2)} ')
```

```python
# 4. How do teams win? (Runs vs Wickets) 
sns.countplot(data=df, x='won_by')
```

```python
df['won_by'].value_counts()
```

Key Player Performances

```python
# 1. Which players received the most "Player of the Match" awards?
df.head()
players = df['player_of_the_match'].value_counts().head(5)
sns.barplot(x=players.values, y=players.index)
plt.title('Man of the match Plot')
```

```python
# 2. Who are the top scorers?
df.head(5)
```

```python
top_scorers = df.groupby('top_scorer')['highscore'].sum()
top_scorer = top_scorers.sort_values(ascending=False).head(5)
top_scorer
```

```python
sns.barplot(x=top_scorer.values, y=top_scorer.index, palette='colorblind')
```

```python
# 3. Which players received the most "Player of the Match" awards?
df.head()
```

```python
# 4. Who are the 10 best bowlers based on bowling figures?
df['highest_wickets'] = (df['best_bowling_figure'].apply(lambda x: x.split('--')[0])).astype(int)
highest_wickets=df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
highest_wickets
```

```python
sns.barplot(x=highest_wickets.values, y=highest_wickets.index, palette='colorblind')
```
