```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

<h2>Distribution Plot For Quantative Data</h2>

We use distribution plots to visualize the distribution of quantitative data.


Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

```python
df = sns.load_dataset('tips')
df
```

```python
df[(df['total_bill']>40)]
```

```python
# check the missing values in (size) column
df['size'].isnull().sum()
```

```python
df['size'].unique()
```

<h3>Hist Plot</h3>
<p>The function is plt.hist(), which creates a histogram to show the distribution of numerical data.</p>

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)

```

```python
df['total_bill'].mean()
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'scatter')
```

A Joint Plot is a visualization used to understand the relationship between two numerical variables along with their individual distributions.

In simple words, a joint plot combines:
1. Scatter plot – Shows the relationship between two variables.
2. Histogram – Shows the distribution of each individual variable.

It is commonly used in data analysis to understand relationships and distributions together.

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'reg')
```

<h3>Pair Plot</h3>
<p>In a Seaborn pair plot, each numerical column is compared with every other numerical column in the DataFrame.</p>

```python
sns.pairplot(df,hue='size',palette='rainbow')
```

```python
sns.boxplot(df['total_bill'])
```

```python
sns.rugplot(df['tip'])
```

<h2>Distribution Plot For Qualitative Data</h2>

>Count Plot </br>
>Bar Plot </br>
>BoxPlot (x-axis (quantative) and y-axis(category))
>Violent Plot </br>
>Strip Plot</br>
>Swarm Plot</br>

```python
df = sns.load_dataset('tips')
df.head()
```

```python
# Here sex, smoker, day, time, size are the qualitative columns.
```

A Count Plot shows how many times each unique category occurs in a column.

```python
# Count Plot of Sex (How many males and females)
sns.countplot(x = df['sex'])
```

```python
# On top of this we can also find the how many males and females are smoker.
sns.countplot(x=df['sex'], hue=df['smoker'])
```

```python
# Make a count plot of unique days.
sns.countplot(df, x='day')
```

A bar plot is used to compare values across different categories.</br>
A bar plot compares a categorical (qualitative) variable with a numerical (quantitative) variable.</br>
For example, suppose we have a dataset showing the total tip for different days.



```python
sns.barplot(data=df, x='day', y='tip', estimator='sum', errorbar=None)
```

```python
# Show the average tip between sex(male and female)
sns.barplot(data=df, x='sex', y='tip', estimator='mean', errorbar=None )
```

```python
# showing the total bill of each day in dataset
sns.barplot(data=df, x='day', y='total_bill',estimator='sum', errorbar=None)
```

<h3>Box Plot</h3>
It is used to understand the distribution of numerical data and indentify the outliers, check spread of data

```python
sns.boxplot(data=df, x='total_bill', y='day')
```

```python
df[(df['total_bill']>30) & (df['day']=='Thur')]
```

```python
sns.violinplot(x='day', y='total_bill', data=df)
```

```python
# Show data bars in form of scatters point
sns.stripplot(x='day', y='total_bill', data=df)
```

```python
sns.swarmplot(x='day', y='total_bill', data=df)
```
